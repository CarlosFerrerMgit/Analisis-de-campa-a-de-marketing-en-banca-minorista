```python
# 1. Importación de librerías para el proyecto

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ajustes de visualizacion para que Pandas no corte columnas ni filas en pantalla
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
```

```python
# 2. Carga y análisis EXCEL

ruta_excel = "../data/raw/customer-details.xlsx" 

# Creamos el objetopara revisar las pestañas existentes
excel_obj = pd.ExcelFile(ruta_excel)

print("Hojas encontradas en el Excel:", excel_obj.sheet_names)
```

```python
# Leemos cada una de las hojas y las guardamos en una lista de DataFrames

lista_hojas = []
for hoja in excel_obj.sheet_names:
    df_temp = pd.read_excel(excel_obj, sheet_name=hoja)
    print(f"Hoja {hoja}: {df_temp.shape[0]} filas y {df_temp.shape[1]} columnas")
    lista_hojas.append(df_temp)

# Concatenamos verticalmente todos los registros en las 3 pestañas del Excel y creamos un nuevo DataFrame
df_customer = pd.concat(lista_hojas, ignore_index=True)

print("\n--- RESUMEN NUEVO DATAFRAME ---")
print(f"Dimensiones totales: {df_customer.shape[0]} filas, {df_customer.shape[1]} columnas")
df_customer.head(3)
```

```python
# 3. Carga y análisis CSV

ruta_csv = "../data/raw/bank-additional.csv"

df_bank = pd.read_csv(ruta_csv)

print(f"Dimensiones de df_bank_additional: {df_bank.shape[0]} filas, {df_bank.shape[1]} columnas")
df_bank.head(3)
```

```python
# 4. Información técnica de columnas, tipos y valores no nulos

df_bank.info()
```

```python
# 5. Conteo de los valores nulos para las diferentes columnas

print("\n--- VALORES NULOS POR COLUMNA EN DF_BANK ---")
print(df_bank.isnull().sum())
```

```python
# 6. Tipos de datos asignados automáticamente por Pandas

print("\n--- TIPOS DE DATOS (DTYPES) ---")
print(df_bank.dtypes)
```

```python
# 7. Cruce de ambos datasets 

# Comprobar si hay identificadores duplicados en cada tabla
print(f"Total registros df_bank: {len(df_bank)} | IDs únicos: {df_bank['id_'].nunique()}")
print(f"Total registros df_customer: {len(df_customer)} | IDs únicos: {df_customer['ID'].nunique()}")

# Comprobar intersección de clientes entre ambas fuentes
ids_bank = set(df_bank['id_'])
ids_cust = set(df_customer['ID'])
interseccion = ids_bank.intersection(ids_cust)

print(f"Clientes en común: {len(interseccion)}")
```

```python
# Eliminar la columna 'Unnamed: 0' si existe antes del cruce de datasets
for df_temp in [df_bank, df_customer]:
    if 'Unnamed: 0' in df_temp.columns:
        df_temp.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# Unir ambas fuentes mediante un INNER JOIN usando left_on y right_on
df = pd.merge(df_bank, df_customer, left_on='id_', right_on='ID', how='inner')

# Como 'id_' e 'ID' contienen la misma información, eliminamos una de ellas para evitar tener este dato duplicado
df.drop(columns=['ID'], inplace=True)

print(f"\nDimensiones tras la unión: {df.shape[0]} filas y {df.shape[1]} columnas.")
df.head(5)
```

```python
# 8. Corrección de formato decimal

columnas_con_coma = ['cons.price.idx', 'cons.conf.idx', 'euribor3m', 'nr.employed']

for col in columnas_con_coma:
    # Convertimos a string 
    # Reemplaza cada coma por un punto
    # pd.to_numeric(..., errors='coerce'): convierte a float.
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', '.'), errors='coerce')

print("Tipos de datos tras la conversión:")
print(df[columnas_con_coma].dtypes)
print("\nPrimeras filas con formato numérico correcto:")
df[columnas_con_coma].head(3)
```

```python
# 9. Limpieza y homogeneización de columnas

columnas_categoricas = ['job', 'marital', 'education', 'contact', 'poutcome', 'y']

for col in columnas_categoricas:
    # Pasamos a minúsculas y quitamos posibles espacios en blanco residuales a los extremos
    df[col] = df[col].astype(str).str.strip().str.lower()
    # Si la cadena convertida era 'nan', la devolvemos a NaN real de NumPy
    df[col] = df[col].replace('nan', np.nan)

print("Valores únicos en 'marital':", df['marital'].unique())
print("Valores únicos en 'education':", df['education'].unique())
print("Valores únicos en 'y':", df['y'].unique())
```

```python
# 10. Fechas

# Diccionario mapeador de meses en español a formato numérico de dos dígitos
meses_es = {
    'enero': '01', 'febrero': '02', 'marzo': '03', 'abril': '04',
    'mayo': '05', 'junio': '06', 'julio': '07', 'agosto': '08',
    'septiembre': '09', 'octubre': '10', 'noviembre': '11', 'diciembre': '12'
}

df['date'] = df_bank['date']

def estandarizar_texto_fecha(fecha_val):
    """
    Convierte '2-agosto-2019' en formato estándar de texto '2019-08-02'.
    """
    if pd.isna(fecha_val) or str(fecha_val).strip().lower() in ('nan', 'nat', 'none', ''):
        return None
    
    partes = str(fecha_val).strip().split('-')
    if len(partes) == 3:
        dia, mes, anio = partes
        mes_num = meses_es.get(mes.lower(), mes)
        # Formateamos YYYY-MM-DD asegurando dos dígitos para el día
        return f"{anio}-{mes_num}-{int(dia):02d}"
    
    return str(fecha_val)

# Convertimos el texto a estándar y forzamos formalmente datetime64 con pd.to_datetime
fechas_limpias = df['date'].apply(estandarizar_texto_fecha)
df['date'] = pd.to_datetime(fechas_limpias, errors='coerce')

# Extraemos año y mes como enteros puros sin decimales (.0) usando 'Int64'
df['contact_year'] = df['date'].dt.year.astype('Int64')
df['contact_month'] = df['date'].dt.month.astype('Int64')

# Comprobaciones de los datos
print("Tipo formal de dato en 'date':", df['date'].dtype)
print("Tipo formal de 'contact_year':", df['contact_year'].dtype)
print(f"Total registros: {len(df)} | Nulos en fecha: {df['date'].isna().sum()}")

df[['date', 'contact_year', 'contact_month']].head(5)
```

```python
# 11. Imputación de valores nulos

# Auditoría previa de nulos
print("--- NULOS ANTES DE LA IMPUTACIÓN ---")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Imputación de variables categóricas de perfil (Moda)
# Rellenamos con la categoría más frecuente de cada columna
df['marital'] = df['marital'].fillna(df['marital'].mode()[0])
df['job'] = df['job'].fillna(df['job'].mode()[0])
df['education'] = df['education'].fillna('university.degree')

# Imputación de 'age' (Mediana condicionada por ocupación)
# Un 'student' o 'retired' tiene edades muy distintas a la media general
df['age'] = df['age'].fillna(df.groupby('job')['age'].transform('median'))
# Si quedase algún caso residual aislado, usamos la mediana global
df['age'] = df['age'].fillna(df['age'].median())

# Imputación de variables macroeconómicas temporales
# Agrupamos por año y mes para asignar el tipo Euribor e IPC real del periodo
df['euribor3m'] = df['euribor3m'].fillna(
    df.groupby(['contact_year', 'contact_month'])['euribor3m'].transform('median')
)
df['euribor3m'] = df['euribor3m'].fillna(df['euribor3m'].median())

df['cons.price.idx'] = df['cons.price.idx'].fillna(
    df.groupby(['contact_year', 'contact_month'])['cons.price.idx'].transform('median')
)
df['cons.price.idx'] = df['cons.price.idx'].fillna(df['cons.price.idx'].median())

# Imputación de variables financieras binarias (default, housing, loan)
df['default'] = df['default'].fillna(0.0)
df['housing'] = df['housing'].fillna(df['housing'].mode()[0])
df['loan'] = df['loan'].fillna(df['loan'].mode()[0])
```

```python
print("\n--- NULOS TRAS LA IMPUTACIÓN ---")
nulos_restantes = df.isnull().sum()[df.isnull().sum() > 0]
if len(nulos_restantes) == 0:
    print("¡Limpieza completada! No quedan valores nulos en el dataset.")
else:
    print(nulos_restantes)
```

```python
# 12. Creación de nuevas variables

# Variable objetivo en formato binario numérico (1: Sí contrató, 0: No contrató)
df['target_numeric'] = df['y'].map({'yes': 1, 'no': 0})

# Total de hijos a cargo en el hogar (Kidhome + Teenhome)
df['total_children'] = df['Kidhome'] + df['Teenhome']

# Indicador de contacto previo:
# En pdays, 999 significa que el cliente nunca antes había sido contactado
df['contacted_before'] = (df['pdays'] != 999).astype(int)

# Segmentación por rangos de edad biológicos y laborales
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 30, 45, 65, 100],
    labels=['<30 joven', '30-45 adulto_joven', '46-65 adulto_senior', '>65 jubilado']
)

print("Nuevas variables generadas exitosamente:")
df[['y', 'target_numeric', 'total_children', 'contacted_before', 'age_group']].head(4)
```

```python
# 11. GUARDAMOS DATOS PROCESADOS
ruta_guardado = "../data/clean/bank_clean.csv"

# Guardamos sin índice para no crear columnas redundantes
df.to_csv(ruta_guardado, index=False)

print(f"Dataset procesado guardado exitosamente en: {ruta_guardado}")
print(f"Volumen final: {df.shape[0]} registros y {df.shape[1]} columnas.")
```
